```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit.quantum_info import DensityMatrix, Statevector , Operator
from qiskit.circuit.library import UCGate
from qiskit.circuit import Parameter
from qiskit.circuit.library import UnitaryGate 
import numpy as np

def AND(total_qubits = 6 , needed_qubits = 3):
    qc = QuantumCircuit(needed_qubits)
    qc.ccx(0 ,1 ,2)
    qc1 = QuantumCircuit(total_qubits)
    for var in range(total_qubits):
        qc1.id(var)

    qc1.compose(qc , qubits = list(range(needed_qubits)) , inplace = True)
    return Operator(qc1).data


    
def XOR(total_qubits=6, needed_qubits=2):
    qc = QuantumCircuit(needed_qubits)
    qc.cx(0, 1)  # CNOT (XOR) on first 2 qubits
    qc1 = QuantumCircuit(total_qubits)
    for i in range(total_qubits):
        qc1.id(i)
    qc1.compose(qc , qubits = list(range(needed_qubits)) , inplace = True)
    return Operator(qc1).data

    
and_matrix = AND()
xor_matrix = XOR()
identity_matrix = np.eye(2**6)
ndentity_matrix = np.eye(2**6)
P_00 = np.kron(np.array([[1,0],[0,0]]), np.array([[1,0],[0,0]]))  # |0><0| ⊗ |0><0|
P_01 = np.kron(np.array([[1,0],[0,0]]), np.array([[0,0],[0,1]]))  # |0><0| ⊗ |1><1|
P_10 = np.kron(np.array([[0,0],[0,1]]), np.array([[1,0],[0,0]]))  # |1><1| ⊗ |0><0|
P_11 = np.kron(np.array([[0,0],[0,1]]), np.array([[0,0],[0,1]]))  # |1><1| ⊗ |1><1|

# Build full 8-qubit unitary: sum over control projectors tensor the 6-qubit unitary
U = (np.kron(P_00, identity_matrix) +
     np.kron(P_01, and_matrix) +
     np.kron(P_10, xor_matrix) +
     np.kron(P_11, identity_matrix))

# Convert big matrix to UnitaryGate
mux_gate = UnitaryGate(U, label="MUX(2ctrl,6target)")

# Create an 8-qubit circuit and append this gate
qc = QuantumCircuit(8)
qc.append(mux_gate, range(8))
qc.draw('mpl')
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
from qiskit.circuit.library import UnitaryGate
import numpy as np
import matplotlib.pyplot as plt
from qiskit.visualization import plot_state_city

# Define AND gate as 6-qubit padded Toffoli
def AND(total_qubits=6, needed_qubits=3):
    qc = QuantumCircuit(needed_qubits)
    qc.ccx(0, 1, 2)
    qc1 = QuantumCircuit(total_qubits)
    for i in range(total_qubits):
        qc1.id(i)
    qc1.compose(qc, qubits=list(range(needed_qubits)), inplace=True)
    return Operator(qc1).data

# Define XOR (CNOT) gate as 6-qubit padded
def XOR(total_qubits=6, needed_qubits=2):
    qc = QuantumCircuit(needed_qubits)
    qc.cx(0, 1)
    qc1 = QuantumCircuit(total_qubits)
    for i in range(total_qubits):
        qc1.id(i)
    qc1.compose(qc, qubits=list(range(needed_qubits)), inplace=True)
    return Operator(qc1).data

# Prepare logic matrices
and_matrix = AND()
xor_matrix = XOR()
identity_matrix = np.eye(2**6)

# Control projectors
P_00 = np.kron([[1, 0], [0, 0]], [[1, 0], [0, 0]])  # |00><00|
P_01 = np.kron([[1, 0], [0, 0]], [[0, 0], [0, 1]])  # |01><01|
P_10 = np.kron([[0, 0], [0, 1]], [[1, 0], [0, 0]])  # |10><10|
P_11 = np.kron([[0, 0], [0, 1]], [[0, 0], [0, 1]])  # |11><11|

# Build full 8-qubit unitary
U = (
    np.kron(P_00, identity_matrix) +
    np.kron(P_01, and_matrix) +
    np.kron(P_10, xor_matrix) +
    np.kron(P_11, identity_matrix)
)

mux_gate = UnitaryGate(U, label="MUX(2ctrl,6target)")

# Create quantum circuit
qc = QuantumCircuit(8)

# Put instruction qubits q0 and q1 into superposition
qc.h(0)
qc.h(1)

# Initialize data qubits (e.g., q2 = 1, q3 = 1 for AND/XOR testing)
qc.x(2)
qc.x(3)

# Apply the MUX gate across all 8 qubits
qc.append(mux_gate, range(8))

# Get the final statevector
state = Statevector.from_instruction(qc)

# Print and plot the statevector
print(state)
plot_state_city(state, figsize=(10, 6))
plt.show()
```
